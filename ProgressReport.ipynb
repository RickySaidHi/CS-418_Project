## Datasets

So far, our project combines four datasets to examine how restaurant inspection outcomes vary across Chicago community areas and how they relate to different attributes.

### Food Inspections

The `Food_Inspections_20261012.csv` dataset comes from the City of Chicago and consists of all of the inspection data up until 2010. This data is the foundation of our analysis. Data points such as inspection IDs, business names, license numbers, facility types, addresses, inspection dates, results, and descriptions of recorded violations are all taken into account when using it.

Inspection outcomes help us examine different violations in food safety across many restaurants, as well as get an idea of the iterations of restaurants over time. A single restaurant can have many inspections, with a list of different violations included in a single record.

### Business Licenses

The `Business_Licenses_20261105.csv` dataset comes from the City of Chicago and includes historical business-license records, rather than only currently active licenses. We are then able to connect the inspection dataset’s License # to the business-license dataset’s LICENSE NUMBER attribute.

This provides the community-area number for each matched license and allows us to connect restaurant inspections to community-area demographics. We can also use this dataset to count the number of restaurant locations in total and by community area.

### Community Data Snapshots

The `CommunityDataSnapshots_Historical_455993172616713182.csv` dataset comes from CMAP and provides all of the demographic and economic data for the Chicago community areas. Using data fields like the community-area number, GEOID, we are able to join across the business-license records and examine inspections by area. Relevant information like the community-area name, total population, and median household income also helps support analysis of these areas.

By using such data, as well as more socioeconomic data, we can better analyze the different trends that could be present in how an environment could impact restaurant food safety.

### Building Footprints

The `Building_Footprints_20261105.csv` dataset comes from the City of Chicago and describes building locations and characteristics. We use this by obtaining building identifiers, address components, and construction years. Matching the building records to restaurant inspection addresses and calculating building age at the time of inspection allows us to analyze whether there is any trend between violations and building age.

## Data Cleaning - `Data_Visuals.ipynb`

We loaded the data and used pandas to select relevant columns and prepare the four different datasets for analysis. Inspection dates were properly converted to datetime values, and license numbers, community-area identifiers, and median household income were all converted to numeric values.

### Restaurant Inspections and Community Areas

We made sure to collect all records classified as Restaurant across all of the available years and remove any duplicates. We also made sure to retain passing inspections without any recorded violations for further analysis.

For the business licenses, we removed anything with nonpositive license numbers, duplicates, and any licenses that had multiple areas to avoid issues with aggregation.

We then joined the inspections to licenses and connected the community areas using inner joins for both, retaining only the matching records.

### Building Ages

To get building ages from our dataset, substantial cleaning was involved to allow proper joining. We made sure to remove commas from building street numbers and exclude any addresses with missing construction years or missing street numbers that would not be usable for conversion. Afterward, we took the address components and combined them into one field, making sure they had the same format so they could be joined properly. Addresses with multiple buildings were excluded in this case, and the remaining records were matched to inspections by exact address.

Following this, we used a left join to preserve the inspections and added a calculation of building age by subtracting the construction year from the given inspection year. This allows for proper analysis as we look at many of the same restaurants over the course of multiple years.

### Inspection Outcomes and Business Counts

For now, we only utilize the difference between the pass and fail outcomes of inspections and calculate the fail counts, pass counts, and failure percentages across the different community areas.

Separately, we filtered historical business licenses to Chicago Retail Food Establishment records. After removing missing identifiers and community areas, we counted each account-and-site combination once and joined the area totals to CMAP income data. These counts represent historical food-establishment locations, not currently operating restaurants.

## ML analysis - `inspection-failure-classification.ipynb`

We used logistic regression to predict whether a restaurant inspection would pass or fail, based on features such as risk category, inspection type, neighborhood, inspection date, and building age. Logistic regression looks for patterns between these features and past inspection outcomes to estimate whether an inspection is likely to fail. To evaluate the model, we compared it with a simple baseline that always predicts the most common result, which is pass. We used four fold cross-validation, training and evaluating the models on different portions of the data, and measured accuracy, precision, recall, and F1-score. This let us compare overall correctness with how well the model identifies failed inspections.

The baseline achieved 73.4% accuracy by predicting that every inspection would pass. It did not identify any failed inspections. Logistic regression had lower accuracy at 59.9%, but it identified 73.5% of actual failures. Of the inspections it predicted would fail, 37.2% actually failed, resulting in a Fail F1-score of 49.4%.

This suggests logistic regression is better at finding failed inspections, but it also produces many false alarms. There is still substantial room to improve the model. We believe it is limited by the information available in our dataset. Building age is missing for many inspections, and our features may not capture important restaurant details, such as food type or past inspection history. Adding more complete data and useful features may improve its predictions.
